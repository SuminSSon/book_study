```python
import tensorflow as tf

# 데이터 정의
x_data = [1, 2, 3]
y_data = [1, 2, 3]

# 변수 정의
W = tf.Variable(tf.random.uniform([1], -1.0, 1.0))
b = tf.Variable(tf.random.uniform([1], -1.0, 1.0))

# X와 Y를 텐서로 정의 (tf.placeholder 대체)
X = tf.constant(x_data, dtype=tf.float32)
Y = tf.constant(y_data, dtype=tf.float32)

# 가설 수식
# 행렬 곱이 아니어서 기본 곱셈 기호 사용
def hypothesis(X):
    return W * X + b

# 손실 함수
def cost_fn():
    return tf.reduce_mean(tf.square(hypothesis(X) - Y))

# 경사 하강법 최적화
optimizer = tf.keras.optimizers.SGD(learning_rate=0.1)

# 훈련 루프
for step in range(100):
    # GradientTape를 사용해 경사 계산
    with tf.GradientTape() as tape:
        cost_val = cost_fn()
    gradients = tape.gradient(cost_val, [W, b])
    optimizer.apply_gradients(zip(gradients, [W, b]))

    # 결과 출력
    if step % 10 == 0:
        print(f"Step {step}, Cost: {cost_val.numpy()}, W: {W.numpy()}, b: {b.numpy()}")

# 테스트
print("\n=== Test ===")
print("X: 5, Y:", hypothesis(5).numpy())
print("X: 2.5, Y:", hypothesis(2.5).numpy())
```
