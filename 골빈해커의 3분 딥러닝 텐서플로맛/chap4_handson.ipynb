```python
import tensorflow as tf
import numpy as np

# 데이터
x_data = np.array(
    [[0, 0], [1, 0], [1, 1], [0, 0], [0, 0], [0, 1]], dtype=np.float32)

y_data = np.array([
    [1, 0, 0],  # 기타
    [0, 1, 0],  # 포유류
    [0, 0, 1],  # 조류
    [1, 0, 0],
    [1, 0, 0],
    [0, 0, 1]
], dtype=np.float32)

# 모델 정의
inputs = tf.keras.Input(shape=(2,))
hidden = tf.keras.layers.Dense(units=3, activation='relu')(inputs)
outputs = tf.keras.layers.Softmax()(hidden)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

# 손실 함수와 옵티마이저 정의
model.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 학습
model.fit(x_data, y_data, epochs=100, verbose=2)

# 예측
predictions = model.predict(x_data)
predicted_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(y_data, axis=1)

# 결과 출력
print('예측값:', predicted_classes)
print('실제값:', true_classes)

# 정확도 계산
accuracy = np.mean(predicted_classes == true_classes)
print('정확도: %.2f' % (accuracy * 100))
```

```python
# [털, 날개]
x_data = np.array(
    [[0, 0], [1, 0], [1, 1], [0, 0], [0, 0], [0, 1]], dtype=np.float32)

# [기타, 포유류, 조류]
y_data = np.array([
    [1, 0, 0],  # 기타
    [0, 1, 0],  # 포유류
    [0, 0, 1],  # 조류
    [1, 0, 0],
    [1, 0, 0],
    [0, 0, 1]
], dtype=np.float32)

# 모델 정의
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=10, activation='relu', input_shape=(2,)),  # 입력층 -> 첫 번째 히든 레이어
    tf.keras.layers.Dense(units=3)  # 출력층
])

# 손실 함수와 옵티마이저 정의
loss_fn = tf.keras.losses.CategoricalCrossentropy(from_logits=True)
optimizer = tf.keras.optimizers.Adam(learning_rate=0.01)

# 모델 컴파일
model.compile(optimizer=optimizer, loss=loss_fn, metrics=['accuracy'])

# 학습
model.fit(x_data, y_data, epochs=100, verbose=2)

# 예측
predictions = model(x_data)
predicted_classes = tf.argmax(predictions, axis=1).numpy()
true_classes = tf.argmax(y_data, axis=1).numpy()

# 결과 출력
print('예측값:', predicted_classes)
print('실제값:', true_classes)

# 정확도 계산
accuracy = np.mean(predicted_classes == true_classes)
print('정확도: %.2f' % (accuracy * 100))
```
